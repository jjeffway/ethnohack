```python
from rdkit import Chem
import pandas as pd


data= pd.read_csv("./datasets/cox2_prepared.csv")

def flat_smiles(smi):
    """Canonical SMILES with stereochemistry stripped."""
    mol = Chem.MolFromSmiles(smi)
    Chem.RemoveStereochemistry(mol)
    return Chem.MolToSmiles(mol)

data["flat"] = data["smiles"].apply(flat_smiles)

# how many compounds share a 2D structure with another entry?
dupes = data["flat"].duplicated(keep=False).sum()
print(f"{dupes} rows share a 2D structure with another row")

# of those, how many disagree on the label?
conflict = data.groupby("flat")["Molecule ChEMBL ID"].nunique()
bad = conflict[conflict > 1]
print(f"{len(bad)} 2D structures carry both labels")
print(data[data["flat"].isin(bad.index)].sort_values("flat")[["smiles", "label"]].head(10))

```
